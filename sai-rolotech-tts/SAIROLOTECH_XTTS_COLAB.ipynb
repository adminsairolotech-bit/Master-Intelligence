```python
# Import and check GPU
from TTS.api import TTS
import torch

print(f'GPU: {torch.cuda.is_available()}')
if torch.cuda.is_available():
    print(f'Device: {torch.cuda.get_device_name(0)}')
```

## Step 2: Load XTTS Model (2-3 minutes first time)

```python
# Initialize voice cloning model
print('Loading XTTS v2 model...')
tts = TTS(model_name='tts_models/multilingual/multi-dataset/xtts_v2', gpu=True)
print('✅ Model ready!')
```

## Step 3: Upload Your Voice Sample

```python
from google.colab import files

# Upload voice sample (3-30 seconds, clear speech)
print('Upload voice sample (.wav or .mp3):')
uploaded = files.upload()
voice_file = list(uploaded.keys())[0]
print(f'✅ Uploaded: {voice_file}')
```

## Step 4: Clone Your Voice!

```python
# HINDI - Your voice saying Hindi
tts.tts_to_file(
    text="नमस्ते, मैं साई रोलोटेक हूं।",
    speaker_wav=voice_file,
    file_path="cloned_hindi.wav",
    language='hi'
)
print('✅ Hindi voice cloned!')
```

```python
# ENGLISH - Your voice saying English
tts.tts_to_file(
    text="Hello, I am SAI Rolotech. We make roll forming machines.",
    speaker_wav=voice_file,
    file_path="cloned_english.wav",
    language='en'
)
print('✅ English voice cloned!')
```

```python
# SPANISH
tts.tts_to_file(
    text="Hola, me llamo SAI Rolotech.",
    speaker_wav=voice_file,
    file_path="cloned_spanish.wav",
    language='es'
)
print('✅ Spanish voice cloned!')
```

## Step 5: Play & Download
